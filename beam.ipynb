```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as smp
```

```python
x,y,k,z,z_0,q_0,A_0 = smp.symbols('x,y,k,z,z_0,q_0,A_0')
```

```python
rsq = x**2+y**2
q = q_0+(z-z_0)
A_z = A_0*q_0/q
phase = smp.exp(-1j*k*((x**2+y**2)/(2*q)))
U = A_z*(phase)

x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

x_range,y_range = np.meshgrid(x,y)

plt.figure(5.8)
plt.imshow()
```
