```python
from scipy.special import j1
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import j1

# parameters (SI units)
lmda = 500e-9          # wavelength: 500 nm
k = 2*np.pi/lmda
w = 1e-3               # aperture radius: 1 mm
z = 1.0                # distance: 1 m
A = np.pi * w**2       # aperture area (if A is something else in your source, change this)

# odd number of points so r = 0 is included
r = np.linspace(-2e-3, 2e-3, 2001)

def I(r):
    u = k * w * r / z
    out = np.ones_like(u)              # limit of 2*J1(u)/u as u -> 0 is 1
    nz = u != 0
    out[nz] = 2 * j1(u[nz]) / u[nz]
    return (A / (lmda * z))**2 * out**2

fig, ax = plt.subplots()
ax.plot(r*1e3, I(r))
ax.set_xlabel("r (mm)")
ax.set_ylabel("I(r)")
ax.set_title("Airy pattern")
plt.show()
```
